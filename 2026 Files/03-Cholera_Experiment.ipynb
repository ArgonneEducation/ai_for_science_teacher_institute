```python
import pandas as pd

# Load the dataset
df = pd.read_csv('https://raw.githubusercontent.com/ArgonneEducation/ai_for_science_teacher_institute/refs/heads/main/data/deaths_by_house.csv')

# Get columns and shape
columns = df.columns.tolist()
num_rows = len(df)

print(f"Columns: {columns}")
print(f"Number of rows: {num_rows}")
df.info()
```

## Gemini Prompt:
Using the data in the csv, and using the definition of "near" being 12.2 or lower, determine the total number of houses with deaths, total without death, total near a sewer, and total far from a sewer along with the total number of houses.

```python
# Define "near"
threshold = 12.2

# Calculate totals
total_houses = len(df)
houses_with_deaths = len(df[df['deaths'] > 0])
houses_without_deaths = len(df[df['deaths'] == 0])
houses_near_sewer = len(df[df['dis_sewers'] <= threshold])
houses_far_sewer = len(df[df['dis_sewers'] > threshold])

print(f"Total Houses: {total_houses}")
print(f"Houses with Deaths: {houses_with_deaths}")
print(f"Houses without Deaths: {houses_without_deaths}")
print(f"Houses near Sewer (<= 12.2): {houses_near_sewer}")
print(f"Houses far from Sewer (> 12.2): {houses_far_sewer}")
```

## Gemini Prompt:
can you build a null contingency table? The top labels are: near sewer; and far from sewer. The left labels are: houses with deaths; and houses without deaths. The far right column should show totals from values you've already calculated, and the bottom row should also show totals that you've already calculated. The values in the middle should be based upon percentage of houses near and far from a sewer. Let me know if you have questions.

```python

# Calculate Expected Values (Null Contingency Table)
# Formula: (Row Total * Column Total) / Grand Total

e_death_near = (houses_with_deaths * houses_near_sewer) / total_houses
e_death_far = (houses_with_deaths * houses_far_sewer) / total_houses
e_nodeath_near = (houses_without_deaths * houses_near_sewer) / total_houses
e_nodeath_far = (houses_without_deaths * houses_far_sewer) / total_houses

print(f"{e_death_near=:.2f}")
print(f"{e_death_far=:.2f}")
print(f"{e_nodeath_near=:.2f}")
print(f"{e_nodeath_far=:.2f}")

# Verification
print(f"Row 1 sum: {e_death_near + e_death_far}")
print(f"Row 2 sum: {e_nodeath_near + e_nodeath_far}")
print(f"Col 1 sum: {e_death_near + e_nodeath_near}")
print(f"Col 2 sum: {e_death_far + e_nodeath_far}")

# Data for the null contingency table
data = {
    'Category': ['Houses with Deaths', 'Houses without Deaths', 'Total'],
    'Near Sewer (<= 12.2)': [e_death_near, e_nodeath_far, e_death_near + e_nodeath_near],
    'Far from Sewer (> 12.2)': [e_death_far, e_nodeath_far, e_death_far + e_nodeath_far],
    'Total': [e_death_near + e_nodeath_near, e_death_far + e_nodeath_far, total_houses]
}

# Create DataFrame
null_table_df = pd.DataFrame(data)
null_table_df = round(null_table_df, 1)

# Save to CSV
null_table_df.to_csv('null_contingency_table.csv', index=False)

print("CSV file 'null_contingency_table.csv' has been created successfully.")
```

### Explanation of Some Code Above
What is a Mask?
In pandas, a mask is a Boolean array (a series of Trueand False values) that is used to filter or select data from a DataFrame based on certain conditions. Masks are created by applying a condition to a column, and they help you focus on specific rows that meet your criteria.  You could think of a mask as an overlay that could be applied to the table that will cover up certain values and leave other values visible.  Masks can be combined with &, which has the same result as overlaying the table with the two masks to see what information remains visible.

## Gemini Prompt:  

can we reject the null hypothesis based upon these contingency tables?

**NOTE:** In the cell below, in order to evaluate the results, Gemini has relied upon an additional library to perform the statistical analysis.  This library is scipy and it is calling upon the statistics package within the library and specifically the chi-squared contingency information for evaluating this type of information

```python
from scipy.stats import chi2_contingency

# Observed contingency table
# Rows: [With Deaths, Without Deaths]
# Columns: [Near Sewer, Far Sewer]
observed = [
    [obs_death_near, obs_death_far],  # Houses with deaths - note that Gemini coded in the values directly instead of using the variables, 
    # but these have been replaced with the variables for clarity and to avoid hardcoding values
    [obs_nodeath_near, obs_nodeath_far]  # Houses without deaths
]

chi2, p, dof, expected = chi2_contingency(observed)

print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"P-value: {p:.4f}")
print(f"Degrees of Freedom: {dof}")
print("\nExpected Frequencies (Calculated by SciPy):")
print(expected)
```

Let's add in code to tell us what the LLM tells us explicitly.  Then we can apply it more generally. Basically it is just going to compare the p-value to a value that we give it (alpha) which is traditionally 0.05 but does not necessarily have to be.



```python
def evaluate_hypothesis(p_value, alpha=0.05):
    """
    Decides whether to reject the null hypothesis based on p-value and alpha.
    """
    print(f"Results Analysis:")
    print(f"  - Calculated p-value: {p_value:.4f}")
    print(f"  - Significance level (alpha): {alpha}")
    print("-" * 35)

    if p_value <= alpha:
        print("DECISION: REJECT the null hypothesis.")
        print("CONCLUSION: The findings are statistically significant.")
    else:
        print("DECISION: FAIL TO REJECT the null hypothesis.")
        print("CONCLUSION: No statistically significant relationship was found.")

# Example usage with the cholera/sewer data:
evaluate_hypothesis(p, alpha=0.05)
```

# How Might We Visualize The Result?
A powerful visualization for deciding if these two populations (near the sewer, far from the sewer) are different based upon deaths is the KDE density plot.  
It provides a distribution for each population so that one can visually evaluate if the distributions are similar.  For the plot, the area under each curve is equal to 1.

Gemini provided the code below to develop this visualization.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter data for houses with and without deaths
deaths_df = df[df['deaths'] > 0]
no_deaths_df = df[df['deaths'] == 0]

# Create the KDE plot
plt.figure(figsize=(10, 6))
sns.kdeplot(deaths_df['dis_sewers'], fill=True, label='Houses with Deaths', color='red')
sns.kdeplot(no_deaths_df['dis_sewers'], fill=True, label='Houses without Deaths', color='blue')

# Add labels and title
plt.title('KDE of Distance to Sewers for Houses With and Without Deaths')
plt.xlabel('Distance to Sewers')
plt.ylabel('Density')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.savefig('kde_distance_to_sewers.png')
print("KDE plot saved as 'kde_distance_to_sewers.png'")
plt.show()
```
